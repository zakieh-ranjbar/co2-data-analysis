```python
# وارد کردن کتابخانه Pandas برای کار با داده‌ها
import pandas as pd
```

```python
# خواندن فایل CSV و ذخیره داده‌ها در یک DataFrame
df = pd.read_csv('../data/co2.csv')
```

```python
# ایجاد متغیر برای ذخیره مجموع مقادیر
total = 0

# محاسبه مجموع 500 مقدار اول ستون out1
for i in range(500):
    total += df['out1'][i]

# نمایش مجموع مقادیر
print("Total:", total)

# محاسبه میانگین 500 مقدار
average = total / 500

# نمایش میانگین محاسبه‌شده
print("Average:", average)
```
